```python
from ultralytics import YOLO
import splitfolders
import os
print("Libraries imported successfully")
```

```python
splitfolders.ratio(
    "PlantVillage",
    output="dataset",
    seed=42,
    ratio=(0.7, 0.2, 0.1)
)
```

```python
from ultralytics import YOLO

model = YOLO("yolov8n-cls.pt")

model.train(
    data="dataset",
    epochs=20,
    imgsz=224
)
```

```python
metrics = model.val()

accuracy = metrics.top1 * 100

print(f"Validation Accuracy: {accuracy:.2f}%")
```
